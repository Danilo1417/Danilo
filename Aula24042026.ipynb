```python
lista_vendas =[1000, 500, 800, 1500, 2000, 2300]
meta = 1200
percentual_extra = 0.1
bonus_extra = 50
for venda in lista_vendas:
  bonus = 0
  if venda >= meta:
    bonus = venda * percentual_extra
    bonus = bonus + bonus_extra

    if venda >= 2000 :
      bonus = bonus + bonus_extra
      bonus +- bonus_extra
  print(f"Venda: {venda} -> bonus : R${bonus}")
```
